```python
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
import skimage
from skimage import io

ROOT = Path('/content/drive/MyDrive/Image processing coursework')
IM_DIR = ROOT / 'Data' / 'Retina' / 'IM'
GT_DIR = ROOT / 'Data' / 'Retina' / 'GT'
# Creates the root direct from my Drive

names = sorted(p.stem for p in IM_DIR.glob('*.jpg'))
# Sorts images
print('scikit-image', skimage.__version__)
print(len(names), names)
# Confirms images sorted and prints for confirmation
```

```python
def load_pair(name):
    # Return the RGB image and its boolean ground truth mask
    image = io.imread(IM_DIR / f'{name}.jpg')
    truth = io.imread(GT_DIR / f'{name}.tif') >= 128
    # Sets the var for images of both types and returns the, to confirm
    # that they are being loaded correctly
    return image, truth

image, truth = load_pair('01_h')
print(image.shape, image.dtype, truth.shape, round(truth.mean(), 4))
# RPrints image with text confirming that name

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].imshow(image)
axes[0].set_title('01_h input')
axes[1].imshow(truth, cmap='gray')
axes[1].set_title('01_h ground truth')
for ax in axes:
    ax.axis('off')
plt.show()
```
